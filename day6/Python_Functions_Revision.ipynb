# Python Functions Revision

Interview-focused revision notes.

## 1. Defining Functions

```
def greet():
    print("Hello")

greet()

def add(a, b):
    return a + b

print(add(10, 20))
```

## 2. Return vs Print

```
def f():
    print(10)

x = f()
print(x)  # None because functions return None by default
```

## 3. Multiple Return Values

```
def calc(a, b):
    return a+b, a-b, a*b

x, y, z = calc(10, 5)
print(x, y, z)
# Actually returns a tuple.
```

## 4. Default Arguments

```
def power(a, b=2):
    return a**b

print(power(5))
print(power(5,3))
```

## 5. Keyword & Positional Arguments

```
def student(name, age):
    print(name, age)

student("Harsh", 21)
student(age=21, name="Harsh")
student("Harsh", age=21)
# student(age=21, "Harsh")  # Invalid
```

## 6. *args

```
def add(*nums):
    print(nums)
    return sum(nums)

print(add(1,2,3,4))
# args is a tuple
```

## 7. **kwargs

```
def info(**data):
    print(data)

info(name="Harsh", age=21)
# kwargs is a dictionary
```

## 8. Scope

```
x = 10

def fun():
    x = 20
    print(x)

fun()
print(x)
```

## 9. global

```
x = 10

def fun():
    global x
    x = 20

fun()
print(x)
```

## 10. nonlocal

```
def outer():
    x = 10
    def inner():
        nonlocal x
        x = 20
    inner()
    print(x)

outer()
```

## 11. Functions are Objects

```
def greet():
    print("Hello")

x = greet
x()
```

## 12. Passing Functions

```
def square(x):
    return x*x

def operate(fun, x):
    return fun(x)

print(operate(square,5))
```

## 13. Lambda Functions

```
square = lambda x: x*x
print(square(5))

add = lambda a,b: a+b
print(add(2,3))
```

## 14. Recursion

```
def fact(n):
    if n == 1:
        return 1
    return n * fact(n-1)

print(fact(5))
```

## 15. Docstrings

```
def add(a,b):
    """Returns sum of two numbers."""
    return a+b

help(add)
```

## 16. Type Hints

```
def add(a:int, b:int) -> int:
    return a+b

print(add(2,3))
```

## 17. Mutable Default Argument (Interview Favorite)

```
def func(lst=[]):
    lst.append(1)
    return lst

print(func())
print(func())

# Correct way:
def safe_func(lst=None):
    if lst is None:
        lst = []
    lst.append(1)
    return lst

print(safe_func())
print(safe_func())
```
